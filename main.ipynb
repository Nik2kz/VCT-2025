```python
import requests
from bs4 import BeautifulSoup
import pandas as pd
import time

events = [
    {"name": "Valorant Champions 2025", "url": "https://www.vlr.gg/event/stats/2283/valorant-champions-2025"},
    {"name": "VCT 2025: Americas Stage 2", "url": "https://www.vlr.gg/event/stats/2501/vct-2025-americas-stage-2"},
    {"name": "VCT 2025: EMEA Stage 2", "url": "https://www.vlr.gg/event/stats/2498/vct-2025-emea-stage-2"},
    {"name": "VCT 2025: Pacific Stage 2", "url": "https://www.vlr.gg/event/stats/2500/vct-2025-pacific-stage-2"},
    {"name": "VCT 2025: China Stage 2", "url": "https://www.vlr.gg/event/stats/2499/vct-2025-china-stage-2"},
    {"name": "VCT 2025: Americas Stage 1", "url": "https://www.vlr.gg/event/stats/2347/vct-2025-americas-stage-1"},
    {"name": "VCT 2025: EMEA Stage 1", "url": "https://www.vlr.gg/event/stats/2380/vct-2025-emea-stage-1"},
    {"name": "VCT 2025: Pacific Stage 1", "url": "https://www.vlr.gg/event/stats/2379/vct-2025-pacific-stage-1"},
    {"name": "VCT 2025: China Stage 1", "url": "https://www.vlr.gg/event/2359/vct-2025-china-stage-1"},
    {"name": "Valorant Masters Toronto 2025", "url": "https://www.vlr.gg/event/stats/2282/valorant-masters-toronto-2025"},
    {"name": "Valorant Masters Bangkok 2025", "url": "https://www.vlr.gg/event/stats/2281/valorant-masters-bangkok-2025"},
    {"name": "VCT 2025: EMEA Kickoff", "url": "https://www.vlr.gg/event/stats/2276/vct-2025-emea-kickoff"},
    {"name": "VCT 2025: Americas Kickoff", "url": "https://www.vlr.gg/event/stats/2274/vct-2025-americas-kickoff"},
    {"name": "VCT 2025: Pacific Kickoff", "url": "https://www.vlr.gg/event/stats/2277/vct-2025-pacific-kickoff"},
    {"name": "VCT 2025: China Kickoff", "url": "https://www.vlr.gg/event/stats/2275/vct-2025-china-kickoff"}
]

def scrape_event_stats(event_name, url):
    headers = {"User-Agent": "Mozilla/5.0"}
    response = requests.get(url, headers=headers)
    
    if response.status_code != 200:
        print(f"Failed to fetch {event_name}: {response.status_code}")
        return None
    
    soup = BeautifulSoup(response.content, "html.parser")
    table = soup.find("table")
    
    if not table:
        print(f"No table found for {event_name}")
        return None
    
    headers_row = table.find("thead").find_all("th")
    col_names = [th.text.strip() for th in headers_row]
    
    rows = []
    for tr in table.find("tbody").find_all("tr"):
        cells = tr.find_all("td")
        row = [td.text.strip() for td in cells]
        if row:
            rows.append(row)
    
    df = pd.DataFrame(rows, columns=col_names)
    df["event"] = event_name
    return df

all_data = []

for event in events:
    print(f"Scraping {event['name']}...")
    df = scrape_event_stats(event["name"], event["url"])
    if df is not None:
        all_data.append(df)
    time.sleep(3)

full_df = pd.concat(all_data, ignore_index=True)
full_df.to_csv("vlr_raw_stats.csv", index=False)
print("Done!", full_df.shape)
```

```python
def get_playoff_match_urls(event_matches_url):
    """Scrape all match URLs from an event playoff page"""
    headers = {"User-Agent": "Mozilla/5.0"}
    response = requests.get(event_matches_url, headers=headers)
    soup = BeautifulSoup(response.content, "html.parser")
    
    match_links = []
    for a in soup.find_all("a", href=True):
        href = a["href"]
        # match URLs are /number/team1-vs-team2-...
        parts = href.strip("/").split("/")
        if len(parts) == 2 and parts[0].isdigit():
            full_url = "https://www.vlr.gg" + href
            if full_url not in match_links:
                match_links.append(full_url)
    
    return match_links

# test on Champions 2025 playoffs
urls = get_playoff_match_urls("https://www.vlr.gg/event/matches/2283/valorant-champions-2025/?series_id=5080")
print(f"Found {len(urls)} matches")
for u in urls:
    print(u)
```
