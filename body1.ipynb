```python
def get_playoff_match_urls(event_matches_url):
    headers = {"User-Agent": "Mozilla/5.0"}
    response = requests.get(event_matches_url, headers=headers)
    soup = BeautifulSoup(response.content, "html.parser")
    
    match_links = []
    for a in soup.find_all("a", href=True):
        href = a["href"]
        parts = href.strip("/").split("/")
        if len(parts) == 2 and parts[0].isdigit():
            full_url = "https://www.vlr.gg" + href
            if full_url not in match_links:
                match_links.append(full_url)
    
    return match_links

urls = get_playoff_match_urls("https://www.vlr.gg/event/matches/2283/valorant-champions-2025/?series_id=5080")
print(f"Found {len(urls)} matches")
for u in urls:
    print(u)
```

```python
def scrape_match(match_url, event_name):
    headers = {"User-Agent": "Mozilla/5.0"}
    response = requests.get(match_url, headers=headers)
    soup = BeautifulSoup(response.content, "html.parser")
    
    rows = []
    
    # get team names
    teams = soup.find_all("div", class_="match-header-link-name")
    if len(teams) < 2:
        return []
    team1 = teams[0].text.strip()
    team2 = teams[1].text.strip()
    
    # find each map section
    map_sections = soup.find_all("div", class_="vm-stats-game")
    
    for map_section in map_sections:
        # skip the "all maps" aggregated section
        map_id = map_section.get("data-game-id", "")
        if map_id == "all":
            continue
        
        # get map name
        map_name_tag = map_section.find("div", class_="map")
        if not map_name_tag:
            continue
        map_name = map_name_tag.text.strip().split("\n")[0].strip()
        
        # get scores to determine winner
        score_tags = map_section.find_all("div", class_="score")
        if len(score_tags) < 2:
            continue
        try:
            score1 = int(score_tags[0].text.strip())
            score2 = int(score_tags[1].text.strip())
            winner = team1 if score1 > score2 else team2
        except:
            continue
        
        # get player stats tables (one per team)
        tables = map_section.find_all("table")
        
        for table_idx, table in enumerate(tables):
            current_team = team1 if table_idx == 0 else team2
            
            for tr in table.find("tbody").find_all("tr"):
                cells = tr.find_all("td")
                if len(cells) < 5:
                    continue
                
                try:
                    player = cells[0].text.strip().split("\n")[0].strip()
                    agent = cells[1].find("img")
                    agent = agent["title"] if agent and agent.has_attr("title") else cells[1].text.strip()
                    acs = cells[3].text.strip()
                    k = cells[4].text.strip().split("\n")[0].strip()
                    d = cells[5].text.strip().split("\n")[0].strip()
                    a = cells[6].text.strip().split("\n")[0].strip()
                    adr = cells[9].text.strip() if len(cells) > 9 else None
                    hs = cells[10].text.strip() if len(cells) > 10 else None
                    
                    rows.append({
                        "event": event_name,
                        "match_url": match_url,
                        "map": map_name,
                        "team": current_team,
                        "opponent": team2 if current_team == team1 else team1,
                        "winner": winner,
                        "player": player,
                        "agent": agent,
                        "ACS": acs,
                        "K": k,
                        "D": d,
                        "A": a,
                        "ADR": adr,
                        "HS%": hs,
                    })
                except:
                    continue
    
    return rows

# test on one match first
test_rows = scrape_match("https://www.vlr.gg/542265/paper-rex-vs-g2-esports-valorant-champions-2025-ubqf", "Valorant Champions 2025")
print(f"Extracted {len(test_rows)} player-map rows")
if test_rows:
    print(test_rows[0])
```

```python
all_match_data = []
failed = []

for i, match in enumerate(all_match_urls):
    print(f"[{i+1}/{len(all_match_urls)}] Scraping {match['url'].split('/')[-1][:50]}...")
    rows = scrape_match(match["url"], match["event"])
    if rows:
        all_match_data.extend(rows)
    else:
        failed.append(match["url"])
    time.sleep(2)

print(f"\nDone! Total rows: {len(all_match_data)}")
print(f"Failed matches: {len(failed)}")

match_df = pd.DataFrame(all_match_data)
match_df.to_csv("vlr_match_data_raw.csv", index=False)
print("Saved to vlr_match_data_raw.csv")
```

```python
match_df = pd.read_csv("vlr_match_data_raw.csv")

for col in ["ACS", "K", "D", "A", "ADR"]:
    match_df[col] = pd.to_numeric(match_df[col], errors="coerce")

match_df["HS%"] = match_df["HS%"].str.replace("%", "").str.strip()
match_df["HS%"] = pd.to_numeric(match_df["HS%"], errors="coerce")

match_df["win"] = (match_df["team"] == match_df["winner"]).astype(int)

match_df["map"] = match_df["map"].str.strip()

print(match_df["map"].value_counts())
print(f"\nShape: {match_df.shape}")
print(f"\nNull counts:\n{match_df.isnull().sum()}")
```

```python
# drop the small number of rows with missing ACS
match_df = match_df.dropna(subset=["ACS"])

# save clean version
match_df.to_csv("vlr_match_data_clean.csv", index=False)

print(f"Final shape: {match_df.shape}")
print(f"Maps: {match_df['map'].nunique()}")
print(f"Players: {match_df['player'].nunique()}")
print(f"Agents: {match_df['agent'].nunique()}")
print(f"Teams: {match_df['team'].nunique()}")
```
